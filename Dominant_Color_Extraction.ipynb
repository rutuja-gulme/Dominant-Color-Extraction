# Dominant Color Extraction

Extracting dominant colors from an image using K-Means clustering and using them to create a segmented image.

```python
import numpy as np
import matplotlib.pyplot as plt
import cv2
```

```python
im = cv2.imread('elephant.jpg') 
im = cv2.cvtColor(im,cv2.COLOR_BGR2RGB)   
original_image=im.shape
print(im.shape)
```

```python
plt.imshow(im)
plt.show()
```

```python
all_pixels = im.reshape((330*500, 3))
print(all_pixels.shape)
```

```python
# Apply K-Means clustering to extract dominant colors
# Use 4 clusters to represent the image with a reduced color palette

from sklearn.cluster import KMeans
```

```python
km = KMeans(n_clusters=4) 
km.fit(all_pixels)   
```

```python
centers = km.cluster_centers_
centers = np.array(centers, dtype='uint8')

print(centers)
```

# Plot all colors are these values

```python
plt.figure(0,figsize=(6,2))

i=1

colors = []
for each_col in centers:
    plt.subplot(1,4,i)
    plt.axis('off')  
    i+=1
    
    colors.append(each_col)
    
    # Color swatch 
    a = np.zeros((100,100,3), dtype='uint8')
    a[:,:,:] = each_col
    plt.imshow(a)

plt.show()
```

# Segmenting Our Original Image

```python
new_img = np.zeros((330*500, 3), dtype='uint8')   # create new same size array for new img  # uint8 - unsigned int

print(new_img.shape)
```

```python
colors
```

```python
km.labels_
```

```python
for i in range(new_img.shape[0]):  
    new_img[i] = colors[km.labels_[i]]
# print(original_image)    
    
new_img = new_img.reshape((original_image))    # reshape it
# print(new_img.shape)
plt.imshow(new_img)   # new_img colorize using 4 colors
plt.show()
```
